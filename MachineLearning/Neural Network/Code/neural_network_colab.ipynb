# Neural Network · Colab 入口
此笔记本使用独立的 Google Colab 云端环境。它拉取公开的 MachineLearningNote 仓库，运行整理版脚本及仓库自带的训练与测试数据。按顺序执行三个代码单元格。如需克隆最新版，请在新的 Colab 运行时从头执行。

```
from pathlib import Path
import subprocess, sys
print('Colab Python:', sys.version.split()[0])
repo_dir = Path('/content/MachineLearningNote')
if not repo_dir.exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/sinuosun/MachineLearningNote.git', str(repo_dir)], check=True)
else:
    print('Repository folder already exists:', repo_dir)
script_path = repo_dir / 'MachineLearning' / 'Neural Network' / 'Code' / 'neural_network.py'
if not script_path.is_file():
    raise FileNotFoundError(f'{script_path} is missing. Use a fresh Colab runtime and rerun from the first cell.')
```

```
import importlib.util, subprocess, sys
needed = {'numpy': 'numpy', 'tensorflow': 'tensorflow'}
missing = [package for module, package in needed.items() if importlib.util.find_spec(module) is None]
if missing:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', *missing], check=True)
else:
    print('NumPy and TensorFlow are already available in this runtime.')
```
